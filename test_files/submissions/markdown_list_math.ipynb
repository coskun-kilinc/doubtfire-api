**List of dangerous math symbols:**
- α: alpha 
- β: beta.  
- γ: gamma.  
- δ: delta.  
- ε: epsilon.  
- ϵ: epsilon.  
- θ: theta.  
- λ: lambda.  
- μ: mu.  
- π: pi.  
- ρ: rho.  
- σ: sigma.  
- τ: tau.  
- φ: phi.  
- ω: omega.  
- Δ: Delta.  
- Σ: Sigma.  
- ≤: leq.  
- ≥: geq.  
- ≠: neq.  
- ≈: approx.  
- ±: pm.  
- ×: times.  
- ÷: div.  
- ∞: infty.  
- ∑: sum.  
- ∏: prod.  
- ∫: int.  
- √: sqrt{}.  
- ∂: partial.  
- ∇: nabla.  

∈: in. 
∉: notin. 
∀: forall. 
∃: exists. 
∧: land. 
∨: lor. 
¬: neg. 
→: to. 
←: leftarrow. 
↔: leftrightarrow. 
⇒: Rightarrow. 
⇔: Leftrightarrow. 
∪: cup. 
∩: cap. 
⊂: subset. 
⊆: subseteq. 
⊃: supset. 
⊇: supseteq. 
∅: emptyset. 


Using these is _dangerous_. What's more, formatting can lead to explosions and sometimes black holes. Use accordingly.
